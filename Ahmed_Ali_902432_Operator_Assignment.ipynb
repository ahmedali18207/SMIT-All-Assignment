## A student is buying a laptop.

    Laptop price: Rs. 85,000
    Student has: Rs. 100,000
    Student gets a discount of Rs. 5,000
    A laptop bag costs Rs. 2,500
    The store has a promotion code "SAVE10".

### Create variables and:

    Calculate the laptop price after discount.
    Add the bag price to get the final bill.
    Calculate the remaining money.
    Check whether the student has enough money.
    Check whether "SAVE10" is present in the promotion code.
    Check whether the final bill is greater than Rs. 80,000.
    Store the result of whether the student has enough money in a variable and check whether it is truthy.

```python
lap=85000
stu_discount=5000
student=100000
lap_bag=2500
promo_code='SAVE10'

lap_price=lap-stu_discount-(lap*0.1)
bill=lap_price+lap_bag
remaining=student-bill
enough_money=student>=bill
promo_present=promo_code=='SAVE10'
bill_limit=bill>80000

print(lap_price)
print(bill)
print(remaining)
print(enough_money)
print(promo_present)
print(bill_limit)
```

### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
math=85
eng=72
python=90
stu_name="Ahmed"
poss_marks=100
req_marks=50

total_marks=math+eng+python
avg_marks=total_marks/3
percentage=(total_marks/(poss_marks*3))*100
math_pass=math>=req_marks
all_pass=math>=req_marks and eng>=req_marks and python>=req_marks
x="A" in stu_name
avg_limit=avg_marks>=80

print(total_marks)
print(avg_marks)
print(percentage,'%')
print(math_pass)
print(all_pass)
print(x)
print(avg_limit)
```
