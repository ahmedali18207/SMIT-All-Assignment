## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []
count = 0

while count < 5:
    num = int(input("Enter marks: "))

    if num == -1:
        break
    
    if num < 0 or num > 100:
        print("Invalid marks")
        continue

    marks.append(num)
    count = count + 1

print("Valid Marks:", marks)

if len(marks) > 0:
    average = sum(marks) / len(marks)
    print("Average =", average)
else:
    print("No valid marks entered")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0

while True:
    print("\n===== BANKING SYSTEM =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Account created successfully!")

    elif choice == "2":
        amount = int(input("Enter deposit amount: "))

        if amount > 0:
            balance = balance + amount
            print("Amount deposited successfully!")
        else:
            print("Invalid amount!")

    elif choice == "3":
        amount = int(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid amount!")
        elif amount > balance:
            print("Insufficient balance!")
        else:
            balance = balance - amount
            print("Amount withdrawn successfully!")

    elif choice == "4":
        print("Your balance is:", balance)

    elif choice == "5":
        print("Thank you for using our bank!")
        break

    else:
        print("Invalid choice! Please try again.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:

    Subject 1: 80
    Subject 2: 75
    Subject 3: 90

    Subject 2: 70
    Subject 3: 65

```python
totals = []

for i in range(1, 4):
    total = 0
    print("Enter marks for Student", i)

    for j in range(1, 4):
        marks = int(input(f"Subject {j}: "))
        total = total + marks

    totals.append(total)
    print("Total marks:", total)
    print("Average:", total / 3)

print("All student totals:", totals)

class_average = sum(totals) / len(totals)
print("Class Average:", class_average)
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []

for day in range(1, 6):
    total = 0
    print("\nDay", day)

    for product in range(1, 4):
        sale = int(input(f"Product {product}: "))

        if sale == 0:
            print("Sale skipped")
            continue

        total = total + sale

    daily_totals.append(total)

print("\nDaily Totals:", daily_totals)

grand_total = sum(daily_totals)
print("Grand Total:", grand_total)
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

Outer loop → for 3 invoices.

Input item price, add to list.

### Example:

```python
invoice_totals = []

for invoice in range(1, 4):
    items = []
    total = 0

    print("\nInvoice", invoice)

    for item in range(1, 5):
        price = int(input(f"Item {item} price: "))
        items.append(price)
        total = total + price

    tax = total * 0.10
    final_total = total + tax

    invoice_totals.append(final_total)

    print("Items:", items)
    print("Subtotal:", total)
    print("Tax (10%):", tax)
    print("Invoice Total:", final_total)

print("\nAll Invoice Totals:", invoice_totals)
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for student in range(1, 6):
    total = 0
    failed = False

    print("\nStudent", student)

    for subject in range(1, 5):
        marks = int(input(f"Subject {subject} marks: "))
        total = total + marks

        if marks < 33:
            print("Failed in Subject", subject)
            failed = True

    percentage = total / 4

    if failed:
        grade = "F"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    print("Total Marks:", total, "/ 400")
    print("Percentage:", percentage, "%")
    print("Grade:", grade)
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
for package in range(1, 4):
    ratings = []
    total = 0

    print("\nPackage", package)

    for feature in range(1, 4):
        rating = int(input(f"Feature {feature} rating (0-5): "))

        if rating == 0:
            print("Rating skipped")
            continue

        ratings.append(rating)
        total = total + rating

    print("Ratings:", ratings)

    if len(ratings) > 0:
        average = total / len(ratings)
        print("Average Rating:", round(average, 2))
    else:
        print("No ratings available")
```
