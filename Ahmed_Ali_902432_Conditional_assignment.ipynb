## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record=input("Criminal record (yes/on)")
no_of_cases=int(input("Enter number of cases:"))
travel_history=input("Travel history flagged (yes/on)")
financial_score=int(input("Enter the financial activity score:"))
last_activity=input("Enter last days activity:")

if criminal_record == "Yes" and cases >3:
    print("High risk")
elif travel_history == "Yes" and financial_score >50:
    print("High risk")
elif last_activity > 365:
    print("Monitor")
elif criminal_record == "No" and financial_score <=50:
    print("Low risk")
else:
    print("Medium risk")
```
