### 7 Write a Python program to calculate compound interest. Use the formula:
A=P*(1+r/n)^n-t

```python
P=float(input('Enter principal amount:'))
r=float(input('Enter annual interest rate:'))
n=float(input('Enter numbers of times interest is compounded per year:'))
t=float(input('Enter time in year:'))

A=P*(1+r/n)**(n*t)

print('Total Amount =',A)
print('Compound Interest =',A-P)
```

### 11 Write a Python program to find the roots of a quadratic equation. Use the quadratic formula:

```python
import math
a=float(input('Enter a value:'))
b=float(input('Enter b value:'))
c=float(input('Enter c value:'))

D = b**2 - 4*a*c

if D>=0:
   x1=(-b + math.sqrt(D))/(2*a)
   x2=(-b - math.sqrt(D))/(2*a)

   print('Root 1=',x1)
   print('Root 2=',x2)
else:
   print('No real roots')
```

### 13 Write a Python program to calculate the gravitational force between two objects.

```python
m1=int(input('Enter the mass of the 1st object in kgs: '))
m2=int(input('Enter the mass of the 2nd object in kgs: '))
r=int(input('Enter the distance in meters between center of m1 and m2: '))
G=6.67e-11
F=G*((m1*m2)/(r**2))
print('The gravitational force between these two objects are: ',F,'meters')
```

### 18 Write a Python program to calculate the future value of an investment.

```python
pv=int(input('Enter the PV (present value) of the investment in pkr: '))
rate=float(input('Enter the annual interest rate on the investment: '))
t=int(input('Enter the time in years for future value: '))
FV=pv*((1+rate)**t)
print('The future value of your investment is: ',FV,'pkr')
```

### 19 Write a Python program to calculate the work done by a force.

```python
force=int(input('Enter the value of force used in Newtons: '))
dist=int(input('Enter the value of distance in meters: '))
theta=float(input('Enter the angle between force and direction of the jmovement in degrees: '))
work=force*dist*math.cos(math.radians(theta))
print('The work done is: ',work,'joules')
```

### 20 Write a Python program to calculate the amount of heat transferred.

```python
m=float(input('Enter the value of the mass in kgs: '))
c=float(input('Enter the specific heat capacity of the material in J/kg.K: '))
Ti=float(input('Enter the initial temperature in Kelvins: '))
Tf=float(input('Enter the final temperature in Kelvins: '))
T=Tf-Ti
Q=m*c*T
print('The amount of heat transferred is: ',Q,'Joules')
```
