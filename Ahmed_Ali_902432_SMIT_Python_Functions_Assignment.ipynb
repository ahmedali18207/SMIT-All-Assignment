**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.

  - Average marks

````python
# Write your code below
def marks_processor(marks):

    total = 0
    highest = marks[0]
    lowest = marks[0]

    for mark in marks:
        total += mark

        if mark > highest:
            highest = mark

        if mark < lowest:
            lowest = mark

    average = total / len(marks)

    return highest, lowest, total, average


# List of student marks
student_marks = [85, 72, 91, 64, 78, 95, 56]

# Receiving multiple return values
high, low, total, avg = marks_processor(student_marks)

# Displaying results
print("Highest Marks:", high)
print("Lowest Marks:", low)
print("Total Marks:", total)
print("Average Marks:", avg)

````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
def process_order(price, quantity, discount_rate, tax_rate):

    # Nested function 1: Calculate subtotal
    def calculate_subtotal():
        return price * quantity

    # Nested function 2: Calculate discount
    def calculate_discount():
        subtotal = calculate_subtotal()
        return subtotal * (discount_rate / 100)

    # Nested function 3: Calculate tax
    def calculate_tax():
        subtotal = calculate_subtotal()
        discount = calculate_discount()

        taxable_amount = subtotal - discount
        return taxable_amount * (tax_rate / 100)

    # Nested function 4: Calculate final amount
    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount()
        tax = calculate_tax()

        final_amount = subtotal - discount + tax
        return final_amount

    return calculate_final_amount()


# Calling the outer function
result = process_order(2000, 3, 10, 18)

print("Final Order Amount: Rs.", result)
````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary
- Allowance
- Deduction
- Tax
- Final salary calculation

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).
- Do **not** write everything inside one function.
- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
# Write your code below

# 1. Employee Information
def employee_info(name, employee_id, department):
    return name, employee_id, department


# 2. Basic Salary
def basic_salary():
    salary = float(input("Enter basic salary: "))
    return salary


# 3. Allowance Calculation
def calculate_allowance(salary):
    house_allowance = salary * 0.15
    medical_allowance = salary * 0.10

    total_allowance = house_allowance + medical_allowance

    return total_allowance


# 4. Deduction Calculation
def calculate_deduction(salary):
    deduction = salary * 0.05
    return deduction


# 5. Tax Calculation
def calculate_tax(gross_salary):

    if gross_salary <= 50000:
        tax_rate = 0

    elif gross_salary <= 100000:
        tax_rate = 0.05

    else:
        tax_rate = 0.10

    tax = gross_salary * tax_rate

    return tax


# 6. Final Salary Calculation
def final_salary(salary, allowance, deduction, tax):

    gross_salary = salary + allowance
    net_salary = gross_salary - deduction - tax

    return gross_salary, net_salary


# MAIN PROGRAM

# Employee details
name = input("Enter employee name: ")
employee_id = input("Enter employee ID: ")
department = input("Enter department: ")

employee = employee_info(name, employee_id, department)

# Salary calculations
salary = basic_salary()

allowance = calculate_allowance(salary)
deduction = calculate_deduction(salary)

gross_salary = salary + allowance

tax = calculate_tax(gross_salary)

gross, net = final_salary(
    salary, allowance, deduction, tax
)

# Final Salary Report
print("\n========== EMPLOYEE PAYROLL REPORT ==========")

print("Employee Name:", employee[0])
print("Employee ID:", employee[1])
print("Department:", employee[2])

print("---------------------------------------------")

print("Basic Salary:       Rs.", salary)
print("Total Allowance:    Rs.", allowance)
print("Gross Salary:       Rs.", gross)
print("Deduction:          Rs.", deduction)
print("Income Tax:         Rs.", tax)

print("---------------------------------------------")

print("NET SALARY:         Rs.", net)

print("=============================================")

````

## Question 28 — Smart Shopping Billing System

**Problem:**  
Build a shopping billing system that can handle multiple products.

**Requirements:**
- Allow the user (or hard-code) multiple products.
- Calculate:
  - Individual product totals
  - Overall subtotal
  - Discount
  - Tax
  - Final bill
- Use **multiple functions** to separate responsibilities.
- At least one function must use `*args` or a flexible-argument design.
- Display a professional-looking bill.

**Focus:** Real-world application + multiple functions + logic building

````python
# Write your code below

# 1. Calculate individual product total
def product_total(price, quantity):
    total = price * quantity
    return total


# 2. Calculate overall subtotal using *args
def calculate_subtotal(*product_totals):
    subtotal = sum(product_totals)
    return subtotal


# 3. Calculate discount
def calculate_discount(subtotal):
    if subtotal >= 10000:
        discount = subtotal * 0.10
    else:
        discount = 0

    return discount


# 4. Calculate tax
def calculate_tax(amount):
    tax = amount * 0.18
    return tax


# 5. Calculate final bill
def final_bill(subtotal, discount, tax):
    bill = subtotal - discount + tax
    return bill


# MAIN PROGRAM

# Product information
products = [
    ("Laptop Bag", 2500, 2),
    ("Mouse", 1200, 1),
    ("Keyboard", 3000, 2)
]

product_totals = []

print("\n========== SMART SHOPPING STORE ==========")
print("\nPRODUCT DETAILS")
print("-" * 55)

# Calculate individual product totals
for name, price, quantity in products:

    total = product_total(price, quantity)

    product_totals.append(total)

    print(name, "| Price:", price,
          "| Quantity:", quantity,
          "| Total:", total)

print("-" * 55)

# Calculate subtotal
subtotal = calculate_subtotal(*product_totals)

# Calculate discount
discount = calculate_discount(subtotal)

# Calculate amount after discount
amount_after_discount = subtotal - discount

# Calculate tax
tax = calculate_tax(amount_after_discount)

# Calculate final bill
bill = final_bill(subtotal, discount, tax)

# Display final bill
print("\n========== FINAL BILL ==========")

print("Subtotal:             Rs.", subtotal)
print("Discount:             Rs.", discount)
print("Amount After Discount: Rs.", amount_after_discount)
print("Tax (18%):            Rs.", tax)

print("-" * 35)

print("FINAL BILL:           Rs.", bill)

print("=" * 35)
print("Thank you for shopping with us!")

````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
# Write your code below

# 1. Student Information
def student_info(name, roll_no, student_class):
    return name, roll_no, student_class


# 2. Accept Subject Marks
def accept_marks():
    subjects = []
    marks = []

    total_subjects = int(input("Enter number of subjects: "))

    for i in range(total_subjects):
        subject = input(f"Enter subject {i+1} name: ")
        mark = float(input(f"Enter marks in {subject}: "))

        subjects.append(subject)
        marks.append(mark)

    return subjects, marks


# 3. Calculate Total Marks Using *args
def calculate_total(*marks):
    total = sum(marks)
    return total


# 4. Calculate Percentage
def calculate_percentage(total, total_possible):
    percentage = (total / total_possible) * 100
    return percentage


# 5. Determine Grade
def calculate_grade(percentage):

    if percentage >= 90:
        grade = "A+"

    elif percentage >= 80:
        grade = "A"

    elif percentage >= 70:
        grade = "B"

    elif percentage >= 60:
        grade = "C"

    elif percentage >= 50:
        grade = "D"

    else:
        grade = "F"

    return grade


# 6. Determine Pass or Fail
def check_result(marks):

    for mark in marks:
        if mark < 40:
            return "FAIL"

    return "PASS"


# 7. Display Final Result Card
def display_result(student, subjects, marks,
                   total, percentage, grade, result):

    print("\n======================================")
    print("          STUDENT RESULT CARD")
    print("======================================")

    print("Student Name:", student[0])
    print("Roll Number:", student[1])
    print("Class:", student[2])

    print("--------------------------------------")
    print("SUBJECT-WISE MARKS")
    print("--------------------------------------")

    for subject, mark in zip(subjects, marks):
        print(subject, ":", mark, "/ 100")

    print("--------------------------------------")
    print("Total Marks:", total, "/", len(subjects) * 100)
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)
    print("Final Result:", result)

    print("======================================")


# MAIN PROGRAM

# Student information
name = input("Enter student name: ")
roll_no = input("Enter roll number: ")
student_class = input("Enter class: ")

student = student_info(name, roll_no, student_class)

# Accept marks
subjects, marks = accept_marks()

# Calculate total
total = calculate_total(*marks)

# Calculate percentage
total_possible = len(subjects) * 100
percentage = calculate_percentage(total, total_possible)

# Calculate grade
grade = calculate_grade(percentage)

# Check pass/fail
result = check_result(marks)

# Display result card
display_result(
    student, subjects, marks,
    total, percentage, grade, result
)

````

## Question 30 — FINAL CHALLENGE: Mini E-Commerce Order System

**Problem:**  
Build a small console-based e-commerce order system using Python functions.

The system should simulate a customer placing an order and produce a professional-looking order summary.

**Suggested functions (you decide the exact design):**
- Customer information
- Product selection / product details
- Price & quantity calculation
- Subtotal
- Discount
- Tax
- Delivery charges
- Final bill
- Order summary display

**Mandatory Requirements:**
- Use multiple `def` functions
- Use arguments (including multiple and default where appropriate)
- Use `return`
- Use conditions and loops
- Use lists and/or dictionaries
- Use `*args` **or** `**kwargs`
- Use **at least one nested function**
- Use **at least one lambda** function where it makes sense
- Keep different tasks in separate functions
- Do **not** write the entire program in one function

**Challenge:**  
Design the program yourself. Decide:
- What functions you need
- What arguments each function needs
- What each function should return
- Where conditions, loops, `*args`/`**kwargs`, nested functions, and lambda are useful

This question tests your **complete understanding** of Python functions.

**Focus:** Complete modular real-world system

````python
# Write your code below

# 1. Customer Information using **kwargs
def customer_info(**kwargs):
    return kwargs


# 2. Display Available Products
def display_products(products):
    print("\n========== AVAILABLE PRODUCTS ==========")

    for product_id, details in products.items():
        print(
            product_id, ".", details["name"],
            "| Price: Rs.", details["price"]
        )


# 3. Product Selection
def place_order(products):
    order = []

    while True:
        product_id = int(input("\nEnter product ID (0 to finish): "))

        if product_id == 0:
            break

        if product_id not in products:
            print("Invalid product ID. Try again.")
            continue

        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Quantity must be greater than zero.")
            continue

        product = products[product_id]

        order.append({
            "name": product["name"],
            "price": product["price"],
            "quantity": quantity
        })

        print("Product added to cart!")

    return order


# 4. Calculate Individual Product Totals and Subtotal
def calculate_subtotal(order):

    # Lambda calculates individual product total
    calculate_item = lambda item: item["price"] * item["quantity"]

    item_totals = list(map(calculate_item, order))

    # *args accepts multiple product totals
    def add_totals(*args):
        return sum(args)

    subtotal = add_totals(*item_totals)

    return item_totals, subtotal


# 5. Calculate Discount
def calculate_discount(subtotal):

    if subtotal >= 50000:
        discount_rate = 0.10

    elif subtotal >= 20000:
        discount_rate = 0.05

    else:
        discount_rate = 0

    discount = subtotal * discount_rate

    return discount


# 6. Calculate Tax
def calculate_tax(amount):
    tax = amount * 0.18
    return tax


# 7. Calculate Delivery Charges
def calculate_delivery(subtotal):

    if subtotal >= 10000:
        delivery = 0
    else:
        delivery = 250

    return delivery


# 8. Generate Final Bill using a Nested Function
def generate_bill(subtotal, discount, tax, delivery):

    # Nested function
    def calculate_payable():
        return subtotal - discount + tax + delivery

    final_amount = calculate_payable()

    return final_amount


# 9. Display Order Summary
def display_summary(customer, order, item_totals,
                    subtotal, discount, tax, delivery, final_amount):

    print("\n" + "=" * 50)
    print("             ONLINE STORE")
    print("             ORDER INVOICE")
    print("=" * 50)

    print("\nCUSTOMER INFORMATION")

    for key, value in customer.items():
        print(key, ":", value)

    print("\nORDER DETAILS")
    print("-" * 50)

    for item, total in zip(order, item_totals):
        print(
            item["name"],
            "| Quantity:", item["quantity"],
            "| Unit Price:", item["price"],
            "| Total:", total
        )

    print("-" * 50)

    print("Subtotal:          Rs.", round(subtotal, 2))
    print("Discount:          Rs.", round(discount, 2))
    print("Tax (18%):         Rs.", round(tax, 2))

    if delivery == 0:
        print("Delivery:          FREE")
    else:
        print("Delivery:          Rs.", delivery)

    print("-" * 50)
    print("FINAL AMOUNT:      Rs.", round(final_amount, 2))

    print("=" * 50)
    print("Thank you for shopping with us!")


# ==========================================
# MAIN PROGRAM
# ==========================================

# Product database
products = {
    1: {"name": "Laptop", "price": 150000},
    2: {"name": "Mouse", "price": 2500},
    3: {"name": "Keyboard", "price": 4500},
    4: {"name": "Monitor", "price": 35000},
    5: {"name": "Headphones", "price": 6000}
}


# Customer information
name = input("Enter customer name: ")
phone = input("Enter phone number: ")
city = input("Enter city: ")

customer = customer_info(
    Name=name,
    Phone=phone,
    City=city
)


# Display products
display_products(products)


# Place order
order = place_order(products)


# Check whether cart is empty
if len(order) == 0:
    print("\nYour cart is empty. No order placed.")

else:

    # Calculate subtotal
    item_totals, subtotal = calculate_subtotal(order)

    # Calculate discount
    discount = calculate_discount(subtotal)

    # Calculate amount after discount
    amount_after_discount = subtotal - discount

    # Calculate tax
    tax = calculate_tax(amount_after_discount)

    # Calculate delivery
    delivery = calculate_delivery(subtotal)

    # Calculate final amount
    final_amount = generate_bill(
        subtotal, discount, tax, delivery
    )

    # Display final order summary
    display_summary(
        customer,
        order,
        item_totals,
        subtotal,
        discount,
        tax,
        delivery,
        final_amount
    )
````
