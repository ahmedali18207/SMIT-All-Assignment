## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
class InvalidAmountError(Exception):
    pass


class InsufficientBalanceError(Exception):
    pass


class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than zero!")

        self.balance += amount

        print("Deposit Successful!")
        print("Deposited Amount:", amount)
        print("Current Balance:", self.balance)

    def withdraw(self, amount):

        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than zero!")

        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient account balance!")

        self.balance -= amount

        print("Withdrawal Successful!")
        print("Withdrawn Amount:", amount)
        print("Remaining Balance:", self.balance)



account = BankAccount("Ahmed", 10000)

try:
    print("Account Holder:", account.name)
    print("Current Balance:", account.balance)

    amount = float(input("\nEnter deposit amount: "))
    account.deposit(amount)

    amount = float(input("\nEnter withdrawal amount: "))
    account.withdraw(amount)

except InvalidAmountError as e:
    print("Transaction Failed!")
    print("Reason:", e)

except InsufficientBalanceError as e:
    print("Transaction Failed!")
    print("Reason:", e)

except ValueError:
    print("Please enter a valid numeric amount!")
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
# Custom Exception Classes

class InvalidProductError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


class InsufficientStockError(Exception):
    pass


class InvalidPaymentError(Exception):
    pass


# E-Commerce Class

class Ecommerce:

    def __init__(self):
        self.products = {
            "laptop": {"price": 100000, "stock": 5},
            "mobile": {"price": 50000, "stock": 10},
            "headphones": {"price": 5000, "stock": 20}
        }

    def place_order(self, product, quantity, payment):

        # Check Product
        if product not in self.products:
            raise InvalidProductError("Product does not exist!")

        # Check Quantity
        if quantity <= 0:
            raise InvalidQuantityError("Quantity must be greater than zero!")

        # Check Stock
        if quantity > self.products[product]["stock"]:
            raise InsufficientStockError("Not enough stock available!")

        # Check Payment
        if payment.lower() not in ["cash", "card", "jazzcash"]:
            raise InvalidPaymentError("Unsupported payment method!")

        # Calculate Bill
        price = self.products[product]["price"]
        total = price * quantity

        # Update Stock
        self.products[product]["stock"] -= quantity

        print("\nOrder Successfully Placed!")
        print("Product:", product)
        print("Quantity:", quantity)
        print("Total Bill:", total)
        print("Payment Method:", payment)
        print("Remaining Stock:", self.products[product]["stock"])


# Main Program

shop = Ecommerce()

try:
    product = input("Enter product name: ").lower()
    quantity = int(input("Enter quantity: "))
    payment = input("Enter payment method: ")

    shop.place_order(product, quantity, payment)

except InvalidProductError as e:
    print("Order Failed!")
    print("Reason:", e)

except InvalidQuantityError as e:
    print("Order Failed!")
    print("Reason:", e)

except InsufficientStockError as e:
    print("Order Failed!")
    print("Reason:", e)

except InvalidPaymentError as e:
    print("Order Failed!")
    print("Reason:", e)

except ValueError:
    print("Please enter a valid numeric quantity!")
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```python
# Custom Exceptions

class StudentNotFoundError(Exception):
    pass

class CourseNotFoundError(Exception):
    pass

class DuplicateRegistrationError(Exception):
    pass

class CourseFullError(Exception):
    pass

class PrerequisiteError(Exception):
    pass


# University Class

class University:

    def __init__(self):

        # Students Data
        self.students = {
            101: {
                "name": "Ahmed",
                "completed_courses": ["Python"],
                "registered_courses": []
            },

            102: {
                "name": "Ali",
                "completed_courses": [],
                "registered_courses": []
            },

            103: {
                "name": "Sara",
                "completed_courses": ["Python"],
                "registered_courses": []
            }
        }

        # Courses Data
        self.courses = {
            "Python": {
                "capacity": 2,
                "registered": [],
                "prerequisite": None
            },

            "Data Science": {
                "capacity": 2,
                "registered": [],
                "prerequisite": "Python"
            },

            "Machine Learning": {
                "capacity": 1,
                "registered": [],
                "prerequisite": "Data Science"
            }
        }

    # Registration Method

    def register_student(self, student_id, course_name):

        # Check Student
        if student_id not in self.students:
            raise StudentNotFoundError("Student does not exist!")

        # Check Course
        if course_name not in self.courses:
            raise CourseNotFoundError("Course does not exist!")

        student = self.students[student_id]
        course = self.courses[course_name]

        # Check Duplicate Registration
        if course_name in student["registered_courses"]:
            raise DuplicateRegistrationError(
                "Student is already registered in this course!"
            )

        # Check Course Capacity
        if len(course["registered"]) >= course["capacity"]:
            raise CourseFullError("Course capacity is full!")

        # Check Prerequisite
        prerequisite = course["prerequisite"]

        if prerequisite is not None:
            if prerequisite not in student["completed_courses"]:
                raise PrerequisiteError(
                    f"Complete {prerequisite} before registering!"
                )

        # Complete Registration
        student["registered_courses"].append(course_name)
        course["registered"].append(student_id)

        print("\nRegistration Successful!")
        print("Student:", student["name"])
        print("Student ID:", student_id)
        print("Course:", course_name)


# Main Program

university = University()

try:
    student_id = int(input("Enter Student ID: "))
    course_name = input("Enter Course Name: ")

    university.register_student(student_id, course_name)

except StudentNotFoundError as e:
    print("Registration Failed!")
    print("Reason:", e)

except CourseNotFoundError as e:
    print("Registration Failed!")
    print("Reason:", e)

except DuplicateRegistrationError as e:
    print("Registration Failed!")
    print("Reason:", e)

except CourseFullError as e:
    print("Registration Failed!")
    print("Reason:", e)

except PrerequisiteError as e:
    print("Registration Failed!")
    print("Reason:", e)

except ValueError:
    print("Please enter a valid Student ID!")
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
# Custom Exception Classes

class PatientNotFoundError(Exception):
    pass


class DoctorNotFoundError(Exception):
    pass


class DoctorUnavailableError(Exception):
    pass


class DuplicateAppointmentError(Exception):
    pass


class InvalidTimeError(Exception):
    pass


# Hospital Class

class Hospital:

    def __init__(self):

        # Patients Data
        self.patients = {
            "P101": "Ahmed",
            "P102": "Ali",
            "P103": "Sara"
        }

        # Doctors Data
        self.doctors = {
            "D05": {
                "name": "Dr. Khan",
                "available_times": ["10:30 AM", "11:30 AM", "12:30 PM"]
            },

            "D06": {
                "name": "Dr. Ahmed",
                "available_times": ["09:30 AM", "01:30 PM", "02:30 PM"]
            }
        }

        # Appointments Data
        self.appointments = []


    # Book Appointment Method

    def book_appointment(self, patient_id, doctor_id, time):

        # Check Patient
        if patient_id not in self.patients:
            raise PatientNotFoundError("Patient does not exist!")

        # Check Doctor
        if doctor_id not in self.doctors:
            raise DoctorNotFoundError("Doctor does not exist!")

        # Validate Time Format
        try:
            valid_time = datetime.strptime(time, "%I:%M %p")
        except ValueError:
            raise InvalidTimeError("Invalid time format! Use 10:30 AM format.")

        time = valid_time.strftime("%I:%M %p")

        doctor = self.doctors[doctor_id]

        # Check Doctor Availability
        if time not in doctor["available_times"]:
            raise DoctorUnavailableError(
                "Doctor is not available at this time!"
            )

        # Check Duplicate Appointment
        for appointment in self.appointments:

            if (appointment["doctor_id"] == doctor_id and
                appointment["time"] == time):

                raise DuplicateAppointmentError(
                    "This doctor already has an appointment at this time!"
                )

        # Book Appointment
        self.appointments.append({
            "patient_id": patient_id,
            "doctor_id": doctor_id,
            "time": time
        })

        print("\nAppointment Booked Successfully!")
        print("Patient:", self.patients[patient_id])
        print("Patient ID:", patient_id)
        print("Doctor:", doctor["name"])
        print("Doctor ID:", doctor_id)
        print("Appointment Time:", time)


# Import Datetime
from datetime import datetime


# Main Program

hospital = Hospital()

while True:

    try:
        print("\n===== HOSPITAL APPOINTMENT SYSTEM =====")

        patient_id = input("Enter Patient ID: ").upper()
        doctor_id = input("Enter Doctor ID: ").upper()
        time = input("Enter Appointment Time: ").upper()

        hospital.book_appointment(patient_id, doctor_id, time)

    except PatientNotFoundError as e:
        print("\nAppointment Failed!")
        print("Reason:", e)

    except DoctorNotFoundError as e:
        print("\nAppointment Failed!")
        print("Reason:", e)

    except DoctorUnavailableError as e:
        print("\nAppointment Failed!")
        print("Reason:", e)

    except DuplicateAppointmentError as e:
        print("\nAppointment Failed!")
        print("Reason:", e)

    except InvalidTimeError as e:
        print("\nAppointment Failed!")
        print("Reason:", e)

    else:
        print("\nThank you for using our hospital system!")

    choice = input("\nBook another appointment? (yes/no): ").lower()

    if choice != "yes":
        print("Hospital System Closed!")
        break
```

```python
# Custom Exception Classes

class InvalidWalletError(Exception):
    pass


class InvalidPINError(Exception):
    pass


class InactiveWalletError(Exception):
    pass


class InvalidAmountError(Exception):
    pass


class InsufficientBalanceError(Exception):
    pass


class TransferLimitError(Exception):
    pass


# Digital Wallet Class

class DigitalWallet:

    def __init__(self):

        self.wallets = {
            "W101": {
                "name": "Ahmed",
                "pin": "1234",
                "balance": 20000,
                "active": True
            },

            "W102": {
                "name": "Ali",
                "pin": "5678",
                "balance": 10000,
                "active": True
            },

            "W103": {
                "name": "Sara",
                "pin": "9999",
                "balance": 15000,
                "active": False
            }
        }

        self.transfer_limit = 50000

    # Transfer Method

    def transfer(self, sender_id, pin, receiver_id, amount):

        # Validate Sender Wallet
        if sender_id not in self.wallets:
            raise InvalidWalletError("Sender wallet does not exist!")

        sender = self.wallets[sender_id]

        # Validate PIN
        if sender["pin"] != pin:
            raise InvalidPINError("Incorrect wallet PIN!")

        # Check Sender Account
        if not sender["active"]:
            raise InactiveWalletError("Sender wallet is inactive!")

        # Validate Receiver Wallet
        if receiver_id not in self.wallets:
            raise InvalidWalletError("Receiver wallet does not exist!")

        receiver = self.wallets[receiver_id]

        # Check Receiver Account
        if not receiver["active"]:
            raise InactiveWalletError("Receiver wallet is inactive!")

        # Validate Amount
        if amount <= 0:
            raise InvalidAmountError(
                "Transfer amount must be greater than zero!"
            )

        # Check Transfer Limit
        if amount > self.transfer_limit:
            raise TransferLimitError(
                "Maximum transfer limit is Rs. 50,000!"
            )

        # Check Balance
        if amount > sender["balance"]:
            raise InsufficientBalanceError(
                "Insufficient wallet balance!"
            )

        # Execute Transfer Only After All Validations
        sender["balance"] -= amount
        receiver["balance"] += amount

        print("\nTransfer Successful!")
        print("Sender:", sender["name"])
        print("Receiver:", receiver["name"])
        print("Transferred Amount: Rs.", amount)
        print("Remaining Balance: Rs.", sender["balance"])

    # Check Balance Method

    def check_balance(self, wallet_id):

        if wallet_id not in self.wallets:
            raise InvalidWalletError("Wallet does not exist!")

        print(
            "Current Balance: Rs.",
            self.wallets[wallet_id]["balance"]
        )


# Main Program

wallet = DigitalWallet()

try:

    print("===== DIGITAL WALLET SYSTEM =====")

    sender_id = input("Enter Sender Wallet ID: ").upper()
    pin = input("Enter Wallet PIN: ")

    receiver_id = input("Enter Receiver Wallet ID: ").upper()

    amount = float(input("Enter Transfer Amount: "))

    wallet.transfer(sender_id, pin, receiver_id, amount)

except InvalidWalletError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except InvalidPINError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except InactiveWalletError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except InvalidAmountError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except InsufficientBalanceError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except TransferLimitError as e:
    print("\nTransfer Failed!")
    print("Reason:", e)

except ValueError:
    print("\nPlease enter a valid numeric amount!")
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```python
# class ProductNotFoundError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


class InsufficientStockError(Exception):
    pass


class RetailInventory:

    def __init__(self):
        self.inventory = {
            "Laptop": {"price": 120000, "stock": 10},
            "Mouse": {"price": 1500, "stock": 20},
            "Keyboard": {"price": 3000, "stock": 15}
        }

    # Add Stock
    def add_stock(self, product, quantity):

        if product not in self.inventory:
            raise ProductNotFoundError("Product inventory mein mojood nahi hai.")

        if quantity <= 0:
            raise InvalidQuantityError("Quantity zero ya negative nahi ho sakti.")

        self.inventory[product]["stock"] += quantity

        print(f"{quantity} units {product} ke stock mein add ho gaye.")

    # View Inventory
    def view_inventory(self):

        print("\n------ INVENTORY ------")

        if not self.inventory:
            print("Inventory empty hai.")
            return

        for product, details in self.inventory.items():
            print(
                f"Product: {product} | "
                f"Price: Rs.{details['price']} | "
                f"Stock: {details['stock']}"
            )

    # Sell Product
    def sell_product(self, product, quantity):

        if product not in self.inventory:
            raise ProductNotFoundError("Product nahi mila.")

        if quantity <= 0:
            raise InvalidQuantityError("Selling quantity positive honi chahiye.")

        if quantity > self.inventory[product]["stock"]:
            raise InsufficientStockError("Itna stock available nahi hai.")

        self.inventory[product]["stock"] -= quantity

        total_price = self.inventory[product]["price"] * quantity

        print(f"\n{quantity} {product} sell ho gaye.")
        print(f"Total Bill: Rs.{total_price}")

    # Search Product
    def search_product(self, product):

        if product not in self.inventory:
            raise ProductNotFoundError("Product nahi mila.")

        details = self.inventory[product]

        print("\n------ PRODUCT FOUND ------")
        print("Name:", product)
        print("Price: Rs.", details["price"])
        print("Stock:", details["stock"])


# Main Program

shop = RetailInventory()

while True:

    print("\n===== RETAIL INVENTORY SYSTEM =====")
    print("1. Add Stock")
    print("2. View Inventory")
    print("3. Sell Product")
    print("4. Search Product")
    print("5. Exit")

    try:

        choice = int(input("\nApni choice select karein: "))

        if choice == 1:

            product = input("Product name: ").strip().title()
            quantity = int(input("Quantity add karein: "))

            shop.add_stock(product, quantity)

        elif choice == 2:

            shop.view_inventory()

        elif choice == 3:

            product = input("Product name: ").strip().title()
            quantity = int(input("Selling quantity: "))

            shop.sell_product(product, quantity)

        elif choice == 4:

            product = input("Product search karein: ").strip().title()

            shop.search_product(product)

        elif choice == 5:

            print("Program successfully close ho gaya.")
            break

        else:

            raise ValueError("Invalid menu choice! 1 se 5 tak select karein.")

    except ValueError as e:

        print("Input Error:", e)

    except ProductNotFoundError as e:

        print("Product Error:", e)

    except InvalidQuantityError as e:

        print("Quantity Error:", e)

    except InsufficientStockError as e:

        print("Stock Error:", e)

    else:

        print("\nOperation successfully complete ho gaya.")

    finally:

        print("Inventory System: Operation finished.")
```

## ### Food Delivery Application

### Goal:
Create a complete console-based food delivery application using exception handling.

### Steps:
- Register customers.
- Select a restaurant.
- Display menu items.
- Take item quantities.
- Validate stock.
- Validate minimum order amount.
- Validate delivery area.
- Process payment.
- Create custom exception classes for business-rule failures.
- Use `try`, specific `except` blocks, `else`, `finally`, and `raise`.
- Prevent invalid operations from partially changing important application data.

### Example:
Customer: Ali
Restaurant: ABC Foods
Item: Burger
Quantity: 2
→ Order placed successfully

```python
#### # ==============================
# CUSTOM EXCEPTION CLASSES
# ==============================

class CustomerNotFoundError(Exception):
    pass


class RestaurantNotFoundError(Exception):
    pass


class ItemNotFoundError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


class InsufficientStockError(Exception):
    pass


class MinimumOrderError(Exception):
    pass


class DeliveryAreaError(Exception):
    pass


class PaymentError(Exception):
    pass


class DuplicateCustomerError(Exception):
    pass


# ==============================
# FOOD DELIVERY APPLICATION
# ==============================

class FoodDelivery:

    def __init__(self):

        self.customers = {}

        self.restaurants = {

            "ABC Foods": {
                "Burger": {"price": 300, "stock": 20},
                "Pizza": {"price": 800, "stock": 10},
                "Fries": {"price": 200, "stock": 30}
            },

            "Food Corner": {
                "Sandwich": {"price": 250, "stock": 15},
                "Biryani": {"price": 400, "stock": 20},
                "Zinger": {"price": 350, "stock": 12}
            }
        }

        self.delivery_areas = [
            "Gulshan",
            "Johar",
            "DHA",
            "North Nazimabad"
        ]

        self.minimum_order = 500

        self.orders = []

    # ==============================
    # REGISTER CUSTOMER
    # ==============================

    def register_customer(self, customer_id, name):

        if not customer_id or not name:
            raise ValueError("Customer ID aur name required hain!")

        if customer_id in self.customers:
            raise DuplicateCustomerError(
                "Customer pehle se registered hai!"
            )

        self.customers[customer_id] = {
            "name": name
        }

        print("Customer successfully registered!")

    # ==============================
    # DISPLAY RESTAURANTS
    # ==============================

    def show_restaurants(self):

        print("\nAVAILABLE RESTAURANTS")

        for restaurant in self.restaurants:
            print("-", restaurant)

    # ==============================
    # DISPLAY MENU
    # ==============================

    def show_menu(self, restaurant):

        if restaurant not in self.restaurants:
            raise RestaurantNotFoundError(
                "Restaurant nahi mila!"
            )

        print(f"\n------ {restaurant} MENU ------")

        for item, details in self.restaurants[restaurant].items():

            print(
                f"{item} | "
                f"Price: Rs.{details['price']} | "
                f"Stock: {details['stock']}"
            )

    # ==============================
    # PLACE ORDER
    # ==============================

    def place_order(
        self,
        customer_id,
        restaurant,
        item,
        quantity,
        area,
        payment_method
    ):

        # 1. Validate Customer

        if customer_id not in self.customers:
            raise CustomerNotFoundError(
                "Customer registered nahi hai!"
            )

        # 2. Validate Restaurant

        if restaurant not in self.restaurants:
            raise RestaurantNotFoundError(
                "Restaurant available nahi hai!"
            )

        # 3. Validate Food Item

        if item not in self.restaurants[restaurant]:
            raise ItemNotFoundError(
                "Food item menu mein nahi hai!"
            )

        # 4. Validate Quantity

        if quantity <= 0:
            raise InvalidQuantityError(
                "Quantity zero se greater honi chahiye!"
            )

        food = self.restaurants[restaurant][item]

        # 5. Validate Stock

        if quantity > food["stock"]:
            raise InsufficientStockError(
                "Restaurant ke paas itna stock available nahi!"
            )

        # 6. Calculate Total

        total = food["price"] * quantity

        # 7. Validate Minimum Order

        if total < self.minimum_order:
            raise MinimumOrderError(
                f"Minimum order Rs.{self.minimum_order} hona chahiye!"
            )

        # 8. Validate Delivery Area

        if area not in self.delivery_areas:
            raise DeliveryAreaError(
                "Is area mein delivery available nahi hai!"
            )

        # 9. Process Payment

        payment_method = payment_method.lower()

        if payment_method not in ["cash", "online"]:
            raise PaymentError(
                "Payment method Cash ya Online hona chahiye!"
            )

        # Online payment simulation
        if payment_method == "online":

            payment_success = input(
                "Online payment successful? (yes/no): "
            ).lower()

            if payment_success != "yes":
                raise PaymentError(
                    "Online payment failed!"
                )

        # ==============================
        # ALL VALIDATIONS PASSED
        # NOW UPDATE IMPORTANT DATA
        # ==============================

        food["stock"] -= quantity

        order = {
            "customer": self.customers[customer_id]["name"],
            "restaurant": restaurant,
            "item": item,
            "quantity": quantity,
            "total": total,
            "area": area,
            "payment": payment_method
        }

        self.orders.append(order)

        print("\n===== ORDER PLACED SUCCESSFULLY =====")
        print("Customer:", order["customer"])
        print("Restaurant:", restaurant)
        print("Food:", item)
        print("Quantity:", quantity)
        print("Total Bill: Rs.", total)
        print("Delivery Area:", area)
        print("Payment:", payment_method)

    # ==============================
    # VIEW ORDER HISTORY
    # ==============================

    def view_orders(self):

        print("\n===== ORDER HISTORY =====")

        if not self.orders:
            print("Abhi koi order place nahi hua.")
            return

        for order in self.orders:

            print("\nCustomer:", order["customer"])
            print("Restaurant:", order["restaurant"])
            print("Item:", order["item"])
            print("Quantity:", order["quantity"])
            print("Total: Rs.", order["total"])


# ==============================
# MAIN PROGRAM
# ==============================

app = FoodDelivery()

while True:

    print("\n===== FOOD DELIVERY SYSTEM =====")
    print("1. Register Customer")
    print("2. Show Restaurants")
    print("3. Show Restaurant Menu")
    print("4. Place Order")
    print("5. View Order History")
    print("6. Exit")

    try:

        choice = int(input("\nEnter your choice: "))

        if choice == 1:

            customer_id = input("Customer ID: ").upper().strip()
            name = input("Customer Name: ").strip()

            app.register_customer(customer_id, name)

        elif choice == 2:

            app.show_restaurants()

        elif choice == 3:

            restaurant = input("Restaurant Name: ").strip().title()

            app.show_menu(restaurant)

        elif choice == 4:

            customer_id = input("Customer ID: ").upper().strip()

            restaurant = input("Restaurant Name: ").strip().title()

            item = input("Food Item: ").strip().title()

            quantity = int(input("Quantity: "))

            area = input("Delivery Area: ").strip().title()

            payment_method = input(
                "Payment Method (Cash/Online): "
            ).strip()

            app.place_order(
                customer_id,
                restaurant,
                item,
                quantity,
                area,
                payment_method
            )

        elif choice == 5:

            app.view_orders()

        elif choice == 6:

            print("Thank you for using Food Delivery System!")
            break

        else:

            raise ValueError("Please select a valid option (1-6)!")

    except ValueError as e:

        print("\nInput Error:", e)

    except CustomerNotFoundError as e:

        print("\nCustomer Error:", e)

    except DuplicateCustomerError as e:

        print("\nRegistration Error:", e)

    except RestaurantNotFoundError as e:

        print("\nRestaurant Error:", e)

    except ItemNotFoundError as e:

        print("\nFood Item Error:", e)

    except InvalidQuantityError as e:

        print("\nQuantity Error:", e)

    except InsufficientStockError as e:

        print("\nStock Error:", e)

    except MinimumOrderError as e:

        print("\nOrder Error:", e)

    except DeliveryAreaError as e:

        print("\nDelivery Error:", e)

    except PaymentError as e:

        print("\nPayment Error:", e)

    else:

        print("\nOperation completed successfully.")

    finally:

        print("Food Delivery System: Operation finished.")
```
